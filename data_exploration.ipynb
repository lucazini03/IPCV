```python
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib.pyplot as plt
plt.style.use('ggplot')
```

```python
# all the csv files in results/
path = 'results/'
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
df_list = [pd.read_csv(os.path.join(path, f)) for f in csv_files]
df = pd.concat(df_list, ignore_index=True)
df = pd.read_csv('results/grid_search_1758896393.6181362.csv')

target = 'score'  # target column
```

```python
sns.pairplot(df[df[target] < 25].drop(columns=['iteration']), hue = target, diag_kind='kde');
```

```python
corr = df.drop(columns=['iteration']).corr()
plt.figure(figsize=(15,10)) # set X and Y size
sns.heatmap(corr, cmap="YlGnBu", annot=True);
```
